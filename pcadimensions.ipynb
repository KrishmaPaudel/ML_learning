```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
X,y = make_blobs(n_samples=500, n_features=5, centers=3, cluster_std=1.5, random_state=46)
```

```python
X
```

```python
scaler= StandardScaler()
X_scaled= scaler.fit_transform(X)
```

```python
X_scaled
```

```python
pca= PCA(n_components=2) # reducing the 5 features into 2 features (5D into 2D)
X_pca= pca.fit_transform(X_scaled)
```

```python
X_pca
```

```python
df_pca=pd.DataFrame(X_pca, columns=['PC1','PC2'])
df_pca['label']= y
```

```python
sns.scatterplot(data=df_pca, x='PC1', y='PC2', hue='label', palette='Set2')
```
