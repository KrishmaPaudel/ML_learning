```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df=sns.load_dataset("titanic")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.drop(['deck','embark_town','who','adult_male','alive','class'],axis=1,inplace=True)
```

```python
df.info()
```

```python
df['age'].fillna(df['age'].mean(), inplace=True)
```

```python
df.info()
```

```python
df.dropna(subset=['embarked'],inplace=True)
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
df['sex']=le.fit_transform(df['sex'])
```

```python
df.head()
```

```python
df['embarked']=le.fit_transform(df['embarked'])  #S=2, C=0, Q=1
```

```python
df.head()
```

```python
df=df.astype(int)
```

```python
df.head()
```

```python
X=df.drop('survived',axis=1)  #axis=1 gives the column
y=df['survived']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report   #for titanic the f1-score test is most prefered because it is concerned with presion(when false positive are costly : the person is actually to survive but is predicted to be dead) and recall both
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
